**任務一**：改成自己的名字、學號、年齡。f 開頭的字串可以用大括號把變數塞進文字裡，很常用。

```python
name = '王小明'      #@param {type:"string"}
sid  = 'B11190000'  #@param {type:"string"}
age  = 19           #@param {type:"integer"}

print('-' * 24)
print(f'姓名：{name}')
print(f'學號：{sid}')
print(f'年齡：{age} 歲，明年 {age + 1} 歲')
print('-' * 24)
```

**任務二**：外圈跑幾次就印幾行。把 rows 改小改大各跑一次看差別。

```python
rows = 5  #@param {type:"slider", min:1, max:9, step:1}

for i in range(1, rows + 1):
    line = ''
    for j in range(1, 10):
        line += f'{i}x{j}={i*j:2d}  '
    print(line)
```

**任務三**：改身高體重看條件判斷。把 18.5、24、27 這三個界線值代進去跑，最容易看出誰管誰。

```python
h_cm = 170.0  #@param {type:"number"}
w_kg = 60.0   #@param {type:"number"}
bmi = w_kg / (h_cm / 100) ** 2

if bmi < 18.5:   r = '體重過輕'
elif bmi < 24:   r = '健康體位'
elif bmi < 27:   r = '體重過重'
else:            r = '肥胖'

print(f'BMI {bmi:.1f}　判讀：{r}')
```

**加碼：字串小實驗**：整句話換掉再跑，看看字串可以怎麼玩。

```python
#@title 字串小實驗（改這一句話）
句子 = '我會用手機寫程式'  #@param {type:"string"}

print('長度：', len(句子), '個字')
print('倒過來：', 句子[::-1])
print('重複三次：', 句子 * 3)
print('✅ 改成別句話再按一次 ▶')
```
