# W03 資料處理基礎
手機開這本請先切成「電腦版網站」（iPhone：網址列 ᴀA → 要求電腦版網站；Android：右上 ⋮ → 電腦版網站），再點每一格左邊的 ▶ 執行。

資料來源是全班填的表單（身高／通勤分鐘／手機時數）。老師公布 ID 就貼進下一格；欄位留空白會自動讀課程附的示範資料。

**任務三之一**：ID 已經填好，直接按 ▶ 就好。表單的欄名是整句題目，太長不好用，所以讀進來之後先改成短欄名，順便把用不到的個資欄（時間戳記、電子郵件）拿掉。

```python
#@title 任務三之一：讀班級資料（直接按 ▶）
SHEET_ID = '1qAGQz2EjJ-B32YLSqjAKW1YC5HXTjaH28KpPQGosCGE'  #@param {type:"string"}

import pandas as pd
BASE = 'https://docs.google.com/spreadsheets/d/'
url = BASE + SHEET_ID + '/export?format=csv'
if not SHEET_ID.strip():   # ←投影片未含，執行所需：沒 ID 就讀示範資料
    url = RAW + 'demo_class.csv'
df = pd.read_csv(url)

# 表單的欄名是整句題目，太長不好寫；用關鍵字對成短欄名，題目改字也不會壞
KEYS = {'身高': '身高', '通勤': '通勤時間', '手機': '手機時數'}
def short_name(c):
    for k, v in KEYS.items():
        if k in str(c):
            return v
    return EN2ZH.get(str(c).strip(), c)   # ←投影片未含：示範資料用的是英文欄名
df.columns = [short_name(c) for c in df.columns]

# 只留要分析的三欄；時間戳記、電子郵件是個資，分析用不到就先拿掉
df = df[[c for c in ['身高', '通勤時間', '手機時數'] if c in df.columns]].copy()
for c in df.columns:
    df[c] = pd.to_numeric(df[c], errors='coerce')

print('✅ 讀到了！幾列幾欄：', df.shape)
df.head()
```

**任務三之二**：五個指令認識資料。把 col 換成另外兩欄再按 ▶，比較三欄的平均與最大值。

```python
print('形狀：', df.shape)
df.info()                # 每欄型態與非空筆數
display(df.describe())   # 數值欄的統計摘要
print(df.isna().sum())   # 每一欄各缺幾筆

col = '手機時數'  #@param ["身高", "通勤時間", "手機時數"]
print(df[col].describe())
```

**任務四**：處理缺失值。method 三個選項各跑一次，記下剩幾筆、平均變成多少。

```python
method = 'median'  #@param ["median", "mean", "drop"]
col = '身高'
print('補之前缺幾筆：', df[col].isna().sum())

if method == 'drop':
    df = df.dropna(subset=[col])
else:
    v = df[col].mean() if method == 'mean' else df[col].median()
    df[col] = df[col].fillna(v)

print('處理後平均：', round(df[col].mean(), 1))
```

**任務四之二**：直接按 ▶ 得到三個答案，把數字連同上面兩格的截圖一起交。

```python
#@title 任務四之二：回答三個問題（直接按 ▶）
print('Q1 全班平均身高：', round(df['身高'].mean(), 1), '公分')
print('Q2 最長通勤時間：', df['通勤時間'].max(), '分鐘')
print('Q3 手機時數中位數：', round(df['手機時數'].median(), 1), '小時')
print('✅ 三個數字＋截圖上傳課程表單，檔名：AI導論_W03_學號_姓名')
```
