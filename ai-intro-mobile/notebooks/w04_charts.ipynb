先在試算表 App 畫一張圖截圖，再回來用 Colab 畫同一份資料，比較兩者差異。

**第 0 格**：跟上週一樣先把班級資料讀進來（沒有 ID 就留空白）。

```python
#@title 第 0 格：讀班級資料（SHEET_ID 空白＝自動用示範資料）
SHEET_ID = ''  #@param {type:"string"}

import pandas as pd
BASE = 'https://docs.google.com/spreadsheets/d/'
url = BASE + SHEET_ID + '/export?format=csv'
if not SHEET_ID.strip():   # ←投影片未含，執行所需：沒 ID 就讀示範資料
    url = RAW + 'demo_class.csv'
df = pd.read_csv(url).rename(columns=EN2ZH)   # ←投影片未含，統一欄名
print('✅ 讀到了！幾列幾欄：', df.shape, '→ 往下開始畫圖')
```

**任務一之二：長條圖**。把通勤時間分成四組，每根柱子是一組的人數。

```python
#@title 長條圖：通勤時間分四組（直接按 ▶）
import matplotlib.pyplot as plt
g = pd.cut(df['通勤時間'], [0, 15, 30, 60, 999],
           labels=['<=15', '16-30', '31-60', '>60'])
cnt = g.value_counts().sort_index()
plt.figure(figsize=(6, 3.5))
plt.bar(cnt.index.astype(str), cnt.values, color='#028090')
plt.xlabel('commute (min)')
plt.ylabel('students')
plt.show()
print('✅ 最高的柱子就是最多人的那一組 → 下一格換直方圖')
```

**任務二：直方圖**。bins 拉到 4 和拉到 30 各截一張，你會發現結論可能不一樣。

```python
bins = 10  #@param {type:"slider", min:4, max:30, step:2}

import matplotlib.pyplot as plt
plt.figure(figsize=(7, 4))
plt.hist(df['手機時數'].dropna(), bins=bins, color='#028090')
plt.xlabel('hours per day')
plt.ylabel('students')
plt.title(f'bins = {bins}')
plt.show()
```

**任務三：散布圖**。點散成一團代表兩欄沒關係，排成一條斜線才是有關係。

```python
x = '通勤時間'  #@param ["身高", "通勤時間", "手機時數"]
y = '手機時數'  #@param ["身高", "通勤時間", "手機時數"]
EN = {'身高': 'height', '通勤時間': 'commute',
      '手機時數': 'phone'}

plt.figure(figsize=(6, 4.5))
plt.scatter(df[x], df[y], s=18, alpha=0.6, color='#02C39A')
plt.xlabel(EN[x])
plt.ylabel(EN[y])
plt.show()
```

**任務四：三張圖並排**。改成你們這組的標題再跑，截圖是本週主要成果。

```python
title = 'Class 115-1 Group A'  #@param {type:"string"}
cols = ['身高', '通勤時間', '手機時數']
names = ['height', 'commute', 'phone']

fig, ax = plt.subplots(1, 3, figsize=(13, 3.6))
for i, c in enumerate(cols):
    ax[i].hist(df[c].dropna(), bins=10, color='#028090')
    ax[i].set_title(names[i])

fig.suptitle(title)
plt.tight_layout()
plt.show()
```

**延伸：箱型圖**。圖上單獨的小圈圈就是離群值——全班誰通勤 120 分鐘？

```python
#@title 延伸：箱型圖抓離群值（換欄位看看）
col = '通勤時間'  #@param ["身高", "通勤時間", "手機時數"]

plt.figure(figsize=(6, 2.6))
plt.boxplot(df[col].dropna(), vert=False)
plt.title(EN[col] + ' (boxplot)')
plt.show()
print('✅ 右邊單獨的小圈圈就是離群值，想想他是誰')
```
