資料是全班用表單交的短評（欄位：短評、情緒）。老師公布試算表 ID 前，留空白就會用課程附的 40 則示範短評。

**任務二之一**：讀全班短評。只改 sid 這一欄；看到表格印出來就成功了。

```python
#@title 任務二之一：讀全班短評（sid 空白＝自動用示範短評）
!pip install -q jieba
import pandas as pd, jieba
base = 'https://docs.google.com/spreadsheets/d/'
sid = ''  #@param {type:"string"}
url = base + sid + '/export?format=csv' if sid.strip() else RAW + 'demo_reviews.csv'  # ←投影片未含
df = pd.read_csv(url).rename(columns={'text': '短評', 'label': '情緒'})  # ←投影片未含
print('✅ 全班共', len(df), '筆短評 → 往下跑斷詞')
df.head()
```

**任務二之二**：中文要先斷詞。比較兩行輸出：加了自訂詞之後，專有名詞就不會被切開。

```python
s = '這家店的滷肉飯真的很好吃'  #@param {type:"string"}
print('加詞前：', ' / '.join(jieba.lcut(s)))

jieba.add_word('滷肉飯')
jieba.add_word('南臺科技大學')
print('加詞後：', ' / '.join(jieba.lcut(s)))
```

**任務三**：訓練情緒分類器。te 是測試集比例（建議 0.1～0.4），越大訓練資料越少，正確率通常會掉。

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
X = [' '.join(jieba.lcut(t)) for t in df['短評']]
te = 0.2  #@param {type:"number"}
a, b, c, d = train_test_split(X, df['情緒'], test_size=te)
v = TfidfVectorizer()
m = LogisticRegression().fit(v.fit_transform(a), c)
print('測試集正確率 =', round(m.score(v.transform(b), d), 3))
```

**任務四**：找出被判錯的句子，討論：是反話、是錯字，還是根本沒學過的詞？（示範資料很乾淨，可能一句都沒錯——把 te 調大成 0.4 再跑就有得討論。）

```python
p = m.predict(v.transform(b))
ans = list(d)
bad = [b[i] for i in range(len(b)) if p[i] != ans[i]]
print('判錯', len(bad), '句，列出前 5 句：')
for s in bad[:5]:
    print(' →', s.replace(' ', ''))
```
