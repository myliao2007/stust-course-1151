# W13 語音辨識：Whisper
手機開這本請先切成「電腦版網站」（iPhone：網址列 ᴀA → 要求電腦版網站；Android：右上 ⋮ → 電腦版網站），再點每一格左邊的 ▶ 執行。

本週不需要 GPU：tiny 模型用 CPU 就跑得動，一段 30 秒的錄音約轉 30～60 秒。錄音檔 m4a、mp3、wav 都可以。

**第 1 格**：安裝並載入模型。安裝加下載第一次要等 1～3 分鐘，進度條在跑就是正常。

```python
#@title 第 1 格：裝套件並載入 Whisper（第一次要等，別切走）
!pip install -q openai-whisper
import whisper, torch
size = 'tiny'  #@param ["tiny", "base", "small"]
print('有沒有 GPU：', torch.cuda.is_available())
model = whisper.load_model(size)
print('✅ 模型載入完成 → 往下跑第 2 格上傳錄音')
```

**第 2 格**：按 ▶ 後會出現「選擇檔案」按鈕，從手機的「檔案」挑你剛錄好的音檔。

```python
#@title 第 2 格：上傳你的錄音（按 ▶ 後點「選擇檔案」）
from google.colab import files   # ←投影片未含，執行所需
up = files.upload()
fname = list(up)[0]
print('✅ 收到檔案：', fname, '→ 往下跑第 3 格轉錄')
```

**第 3 格**：轉錄。看到中文跑出來就成功了，把這段文字截圖，等一下要比對。

```python
#@title 第 3 格：轉錄你自己的錄音（tiny 約 30～60 秒）
r = model.transcribe(fname, language='zh', fp16=False)
print('✅ Whisper 聽到的是：')
print(r['text'])
```

**備用格**：上傳一直失敗才用這格——先把錄音存到「我的雲端硬碟 / AI115」資料夾再跑。

```python
#@title 備用格：改從雲端硬碟讀錄音
from google.colab import drive
drive.mount('/content/drive')
name = 'my_voice.m4a'  #@param {type:"string"}
p = '/content/drive/MyDrive/AI115/' + name
r = model.transcribe(p, language='zh', fp16=False)
print(r['text'])
```

**第 4 格**：把手機語音輸入打出來的那段字貼進 phone 欄位，跟 Whisper 的結果比相似度。

```python
#@title 第 4 格：比較手機聽寫與 Whisper
phone = '手機打出來的那一段'  #@param {type:"string"}
w = r['text']
import difflib
sim = difflib.SequenceMatcher(None, phone, w).ratio()
print('兩段文字相似度 =', round(sim, 3))
print('手機', len(phone), '字　Whisper', len(w), '字')
```
