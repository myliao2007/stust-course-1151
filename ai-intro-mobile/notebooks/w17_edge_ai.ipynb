### 計時頁在做什麼：關鍵這幾行

老師計時頁的核心，你只要改 `N` 那一行再按開始（這是計時頁的 JavaScript，不是這本筆記本的程式）：

```javascript
const N = 10;            // ← 改這裡：要量幾次
const t = [];
for (let i = 0; i < N; i++) {
  const s = performance.now();
  await model.predict(webcam.canvas);   // 本地推論
  t.push(performance.now() - s);
}
const avg = t.reduce((a, b) => a + b) / N;
console.log('平均', avg.toFixed(1), 'ms');
```

**計時只包住 `predict` 那一行，量的才是純推論時間。**

## 任務二：把十次延遲統計起來

**怎麼做**：把中括號裡的十個數字換成你量到的，按 ▶ 就好。

第一次通常最慢（模型還在暖機、記憶體還在配置），這是正常的，**不要把它刪掉**。
連續量十次後手機會微微發熱，這時候會降頻，所以**中位數通常比平均值更能代表**。

**延遲紀錄表（十次都要記）**

| 要記的項目 | 你量到的值 | 備註 |
|---|---|---|
| 第 1 次 | 　ms | 第一次通常最慢，模型還在暖機 |
| 第 2 到 10 次 | 　ms | 十個數字全部抄下來，不要只抄平均 |
| 十次平均 | 　ms | 由第 1 格程式算出來 |
| 中位數 | 　ms | 手機發熱降頻時，中位數比平均可靠 |
| 最快與最慢 | 　ms | 差距太大就說明是什麼原因造成的 |
| 一秒能跑幾次 | 　次 | 1000 除以平均延遲 |
| 換成雲端要多久 | 　ms | 用第 2 格程式算，並和本地比幾倍 |

````python
#@title 第 1 格：把十次延遲統計起來
ms = [42, 38, 55, 40, 39, 41, 44, 37, 48, 40]  # ← 改成你量到的
ms = sorted(ms)
n = len(ms)
avg = sum(ms) / n
print('十次平均 =', round(avg, 1), 'ms')
print('中位數 =', ms[n // 2], 'ms')
print('最快 / 最慢 =', ms[0], '/', ms[-1], 'ms')
print('一秒最多可以跑', int(1000 / avg), '次')
````

## 任務二（下）：如果改送雲端，要多久？

**怎麼做**：把 `local_ms` 換成你在上一格量到的平均，其他先用預設值。

單位換算要記住：**1 Mbps 就是每毫秒 1 kb**，所以把 kb 除以 Mbps 直接得到毫秒。
雲端一次來回 ＝ 上傳時間 ＋ 網路來回 ＋ 伺服器運算。

跑完把「是本地的幾倍」抄進紀錄表最後一列。

````python
#@title 第 2 格：如果改送雲端要多久
photo_kb = 200   #@param {type:"integer"}
up_mbps = 4      #@param {type:"number"}
rtt_ms = 50      #@param {type:"integer"}
server_ms = 30   #@param {type:"integer"}
local_ms = 40    #@param {type:"number"}
up = photo_kb * 8 / up_mbps   # kb 除以 Mbps 就是毫秒
cloud = up + rtt_ms + server_ms
print('上傳花了', round(up), 'ms')
print('雲端一次來回 =', round(cloud), 'ms')
print('是本地的', round(cloud / local_ms, 1), '倍')
````
