## 任務二：for 迴圈印出乘法表

外層跑 2 到 9 段，內層跑 1 到 9。兩層都是 `for`，內層縮排要比外層再多一層。

**會看到**：印出 2 到 9 共八段，最後一行是 `9 x 9 = 81`。

````python
n = 9   # 要印到幾的乘法表

for i in range(2, n + 1):          # 外層：第幾段，range 含頭不含尾所以寫 n + 1
    print(f'--- {i} 的乘法 ---')
    for j in range(1, 10):         # 內層：1 到 9，同樣含頭不含尾
        print(f'{i} x {j} = {i * j}')   # f-string 裡的大括號會算出結果再填進去
````

**換你改**：把 `n` 改成 12，而且**只印偶數段**（2、4、6…）。
提示：`range` 還有第三個參數，代表「每次跳多少」。

## 任務三：if/elif 做成績分級

**順序是關鍵**。由上往下比，先擋掉大的，後面就不用再寫「而且小於 90」。

**會看到**：五個分數依序是 `A B C 不及格 C`；60 分判成 C。

````python
scores = [95, 82, 71, 58, 60]

for s in scores:
    if s >= 90:           # 先擋 90 以上
        grade = 'A'
    elif s >= 80:         # 走到這裡表示已經小於 90，所以只要再比 80
        grade = 'B'
    elif s >= 60:         # 同理，這裡已經小於 80
        grade = 'C'
    else:                 # 以上都不成立
        grade = '不及格'
    print(s, grade)
````

**換你改**：現在 60 分是 C。改成 **60 到 69 給 D、70 到 79 才是 C**，
跑出來第五個分數要變成 `D`。提示：多一個 `elif`，而且順序一樣要由大到小。

## 任務四：寫一個有回傳值的函式

`def` 開頭、冒號結尾、內容縮排；`return` 把結果交回給呼叫的人。

**會看到**：`22.0` 與 `17.6`；對不上就檢查有沒有先把公分換成公尺。

````python
def bmi(h_cm, w_kg):
    m = h_cm / 100         # 公分換算成公尺，172 → 1.72
    value = w_kg / (m ** 2)   # BMI ＝ 體重 ÷ 身高平方，** 是次方
    return value              # 沒寫 return 的話，呼叫的人只會拿到 None

print(round(bmi(172, 65), 1))   # 應該印出 22.0
print(round(bmi(160, 45), 1))   # 應該印出 17.6
````

**換你改**：讓 `bmi()` 除了數值以外，再回傳一句判讀
（未滿 18.5 是「過輕」、18.5 到 24 是「正常」、超過 24 是「過重」）。
提示：`return value, text` 可以一次回傳兩個東西。

**想一想**：把 `return` 那一行刪掉再跑一次，印出來會是 `None`——
因為函式算完了卻沒有把結果交出來。

## 任務五（進階）：while 猜數字

次數事先不知道要跑幾次，這種就用 `while`。
`while True` 表示一直做下去，**靠 `break` 跳出來**。

**會看到**：猜中會印出你猜了幾次。輸入框出現在輸出區下方，點一下再打字。
萬一停不下來，用「執行階段 → 中斷執行」。

````python
import random
answer = random.randint(1, 100)
count = 0

while True:                    # 一直猜，直到 break 為止
    g = int(input('猜一個 1 到 100 的整數：'))
    count += 1
    if g > answer:
        print('太大了')
    elif g < answer:           # 不大於，也不等於，那就是太小
        print('太小了')
    else:
        print('答對了，你猜了', count, '次')
        break                  # 跳出整個 while
````

**換你改**：讓它**最多只能猜 7 次**，超過就直接公布答案並結束。
提示：`while count < 7:`，迴圈結束後再補一行 `print`。
