### matplotlib 指令速查（今天全部會用到）

| 要做什麼 | 指令 | 備註 |
|---|---|---|
| 開一張新圖 | `plt.figure(figsize=(6, 4))` | 單位是英吋，先開再畫 |
| 長條圖 | `plt.bar(x, height)` | `x` 是類別名稱的清單 |
| 直方圖 | `plt.hist(data, bins=10)` | `data` 記得先 `dropna()` |
| 散布圖 | `plt.scatter(x, y, alpha=0.6)` | `alpha` 是透明度，0 到 1 |
| 軸名與標題 | `plt.xlabel / ylabel / title` | 中文會變空格，一律用英文 |
| 加一條水平線 | `plt.axhline(y=3, ls='--')` | 任務五畫平均線要用 |
| 存成檔案 | `plt.savefig('a.png', dpi=150)` | 一定要寫在 `plt.show()` 前面 |
| 顯示出來 | `plt.show()` | `show` 之後畫布會被清空 |

## 任務四：2×2 子圖排版，存成 PNG 下載

**存檔要寫在 `show` 前面**，寫反了會存出一張空白圖。

**會看到**：四格圖（三張直方圖＋一張散布圖），左側檔案面板出現 `w04.png`，
右鍵下載 —— 這是本週主要成果。

````python
cols = ['height_cm', 'commute_min', 'phone_hours']

fig, ax = plt.subplots(2, 2, figsize=(10, 7))
axs = ax.____()                 # ← ① 把 2x2 攤成一維
for i, c in enumerate(cols):
    axs[i].hist(df[c].dropna(), bins=10, color='#028090')
    axs[i].set_title(c)
axs[3].scatter(df['commute_min'], df['phone_hours'], s=14)
axs[3].set_title('commute vs phone')

fig.suptitle('Class 115-1  41B12345')   # 改成你的學號
plt.tight_layout()
plt.____('w04.png', dpi=150)    # ← ② 存成 PNG 檔
plt.show()
````

### 常見狀況排除

| 狀況 | 原因 | 怎麼處理 |
|---|---|---|
| 存出來是空白圖 | `savefig` 寫在 `show` 後面 | 把 `savefig` 移到 `show` 上面一行 |
| 中文標題變空格 | Colab 沒有中文字型 | 軸名與標題一律用英文 |
| 圖一張疊一張 | 沒有先 `plt.figure()` | 每張圖前面都開一張新的 figure |
| 找不到 `w04.png` | 檔案面板沒有重新整理 | 點面板上的重新整理圖示 |
| 下載被 Chrome 擋 | 被當成多重下載 | 點網址列右邊圖示，選一律允許 |
| 圖很糊 | dpi 太低 | `savefig` 加 `dpi=150` 或 200 |
| `NameError: df` | 重新啟動後沒重跑 | 從第 1 格讀資料那格依序重跑 |
| 下課後圖不見 | 存在執行階段暫存區 | 一定要下載到電腦或存回雲端硬碟 |
