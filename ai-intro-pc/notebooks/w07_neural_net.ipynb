## 任務四的資料：全班自己填的看電影決策

跟課堂例題同一組特徵，只是這次**讓網路自己去學權重**。

- **資料怎麼來**：課前 Google 表單，每人填 3 筆
- `rating`：這部片評價好不好，0 到 1 的小數
- `price`：票價貴不貴，0 是便宜、1 是很貴
- `withfriend`：有沒有人一起去，填 0 或 1
- `go`：你最後有沒有去，0 或 1，**這欄是答案**

老師會把彙整好的 `movie.csv` 放到課程雲端硬碟。先下載到電腦，
下一格會用 `files.upload()` 傳進 Colab。

> 課後補做、或還沒收齊表單時，把 `USE_UPLOAD` 改成 `False`，
> 程式會自己產生一份同樣欄位的示範資料，觀念與流程完全一樣。

````python
import pandas as pd
import numpy as np                 # ←投影片未含，執行所需
USE_UPLOAD = True   # 老師發了 movie.csv 就維持 True；課後補做改成 False

if USE_UPLOAD:
    from google.colab import files
    up = files.upload()          # 選 movie.csv（課程雲端硬碟）
    df = pd.read_csv(list(up.keys())[0])
else:                            # ←投影片未含，執行所需：沒有 movie.csv 時的示範資料
    rng = np.random.default_rng(7)
    n = 126
    df = pd.DataFrame({'rating': rng.integers(0, 11, n) / 10,
                       'price': rng.integers(0, 11, n) / 10,
                       'withfriend': rng.integers(0, 2, n)})
    z2 = (0.5 * df['rating'] - 0.8 * df['price']
          + 0.6 * df['withfriend'] - 0.3)
    df['go'] = (z2 + rng.normal(0, 0.05, n) > 0).astype(int)

print(df.shape)              # 例如 (126, 4)
df.head()
````

## 任務四：建一個小網路，讓它自己學權重

**寫對了會看到**：`測試正確率 = 0.895` 之類的數字，超過 0.8 就算合理。

````python
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split

X = df[["rating", "price", "withfriend"]]
y = df["go"]
Xtr, Xte, ytr, yte = train_test_split(
    X, y, test_size=0.3, random_state=42)

clf = MLPClassifier(hidden_layer_sizes=(4,), max_iter=2000,
                    random_state=42)
clf.____(Xtr, ytr)           # ← 自己寫：訓練用哪個方法
print("測試正確率 =", round(clf.____(Xte, yte), 3))
````

## 進階任務：改隱藏層大小，看學習曲線

隱藏層太小學不起來，太大在小資料上容易背答案。這一格不用填空，直接跑。

**會看到**：四條 loss 曲線疊在一起，看哪一條降得快、降得低；
另外印出四個測試正確率。

````python
import matplotlib.pyplot as plt

for h in [2, 4, 8, 16]:
    m = MLPClassifier(hidden_layer_sizes=(h,),
                      max_iter=2000, random_state=42)
    m.fit(Xtr, ytr)
    plt.plot(m.loss_curve_, label="hidden=" + str(h))
    print(h, round(m.score(Xte, yte), 3))

plt.xlabel("iteration")
plt.ylabel("loss")
plt.legend()
plt.savefig("w07_curve.png", dpi=150)
plt.show()
````
