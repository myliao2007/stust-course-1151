#### 參考程式（投影片第 39 頁，`tensor_shape.py`）

> 投影片說明：九成的執行期錯誤都是形狀不合。養成習慣：每寫完一層就 print 一次 shape，比事後回頭猜快得多。

```python
import torch
x = torch.arange(24).reshape(2, 3, 4)   # (2,3,4)
print(x.view(2, 12).shape)              # view 需要記憶體連續
print(x.reshape(-1).shape)              # -1 由其餘維度推算
print(x.permute(0, 2, 1).shape)         # 交換維度 -> (2,4,3)

# unsqueeze / squeeze：增刪長度為 1 的維度
y = torch.randn(3)
print(y.unsqueeze(0).shape)             # (1,3)，補上 batch 維

# broadcasting：從最後一維往前對齊，相同或其一為 1 即可
a = torch.ones(3, 1)
b = torch.ones(1, 4)
print((a + b).shape)                    # (3,4)

# 矩陣乘法用 @ 或 matmul；逐元素相乘用 *
m = torch.randn(2, 3) @ torch.randn(3, 5)
print(m.shape)                          # (2,5)
```

#### ✏️ 動手做

步驟 1、2 就是上面的參考寫法。下一格是數折點的小工具，看不懂沒關係，會用就好。

```python
# 小工具：數一條輸出曲線有幾個折點（斜率突然改變的地方）
import copy

def count_kinks(net, x):
    net, x = copy.deepcopy(net).double(), x.double()   # 用 float64 算，小數誤差才不會被當成折點
    with torch.no_grad():
        y = net(x).squeeze()
    slope = torch.diff(y) / torch.diff(x.squeeze())
    jump = (torch.diff(slope).abs() > 1e-6).nonzero().squeeze(1).tolist()
    groups = [i for k, i in enumerate(jump) if k == 0 or i - jump[k - 1] > 1]
    return len(groups)

x = torch.linspace(-3, 3, 400).unsqueeze(1)
for n in (1, 3, 10):
    torch.manual_seed(n)                 # 和參考寫法同一個種子，數的就是上面那三條
    net = nn.Sequential(nn.Linear(1, n), nn.ReLU(), nn.Linear(n, 1))
    print(f'n = {n:2d}：看到 {count_kinks(net, x)} 個折點（最多 {n} 個）')
```

```python
# ✏️ 步驟 4：兩層（每層 3 個 ReLU）和一層 6 個比一比，各換 5 個亂數種子
def one_layer(seed):
    torch.manual_seed(seed)
    return nn.Sequential(nn.Linear(1, 6), nn.ReLU(), nn.Linear(6, 1))

def two_layers(seed):
    torch.manual_seed(seed)
    return nn.Sequential(nn.Linear(1, 3), nn.ReLU(),
                         nn.Linear(3, 3), nn.ReLU(),
                         nn.Linear(3, 1))

fig, axes = plt.subplots(2, 5, figsize=(16, 5.5), sharex=True)
for j, seed in enumerate(range(5)):      # ✏️ 可以換別的種子，例如 range(10, 15)
    for i, make in enumerate((one_layer, two_layers)):
        net = make(seed)
        with torch.no_grad():
            axes[i, j].plot(x, net(x))
        name = '1 layer x 6' if i == 0 else '2 layers x 3'
        axes[i, j].set_title(f'{name}, seed {seed}: {count_kinks(net, x)} kinks', fontsize=9)
plt.tight_layout(); plt.show()

# 五個種子不夠看的話，一次試 100 個，記下各自看到的最多折點
best1 = max(count_kinks(one_layer(s), x) for s in range(100))
best2 = max(count_kinks(two_layers(s), x) for s in range(100))
print('100 個種子裡看到最多：一層 6 個 →', best1, '個折點；兩層各 3 個 →', best2, '個折點')
```

#### 老師補充：把權重刻意排好，看「折痕相乘」

隨機權重很少剛好排得整整齊齊，所以上面兩層不一定每次都贏。下面這格把權重**刻意排好**：
每層只用 2 個 ReLU，做成一個「山形」（0 → 1 → 0），再一層一層疊上去。
數數看每多一層，折點變成幾個；再和「一層放同樣多個神經元，最多幾個折點」比一比。

```python
# 每層 2 個 ReLU 做成山形：t(u) = 2·ReLU(u) − 4·ReLU(u − 0.5)，u 在 0～1 之間
def mountain(layers):
    mods = [nn.Linear(1, 2), nn.ReLU()]
    for _ in range(layers - 1):
        mods += [nn.Linear(2, 2), nn.ReLU()]
    mods += [nn.Linear(2, 1)]
    net = nn.Sequential(*mods)
    with torch.no_grad():                # 手動設權重，不用訓練
        net[0].weight.copy_(torch.tensor([[1.], [1.]]))
        net[0].bias.copy_(torch.tensor([0., -0.5]))
        for m in net[2:-1]:
            if isinstance(m, nn.Linear):
                m.weight.copy_(torch.tensor([[2., -4.], [2., -4.]]))
                m.bias.copy_(torch.tensor([0., -0.5]))
        net[-1].weight.copy_(torch.tensor([[2., -4.]]))
        net[-1].bias.zero_()
    return net

u = torch.linspace(0, 1, 2001).unsqueeze(1)
fig, axes = plt.subplots(1, 4, figsize=(16, 3.2))
for k, ax in zip((1, 2, 3, 4), axes):
    net = mountain(k)
    with torch.no_grad():
        ax.plot(u, net(u))
    ax.set_title(f'{k} layer(s) x 2 ReLU ({2 * k} neurons): {count_kinks(net, u)} kinks', fontsize=9)
plt.tight_layout(); plt.show()
# 一層放 2k 個神經元，最多 2k 個折點；疊 k 層、每層 2 個，折點是 2^k - 1 個
```
