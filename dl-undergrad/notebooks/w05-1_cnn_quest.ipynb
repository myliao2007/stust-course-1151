1. **每題截圖 × 9**：第 1 題到第 9 題，每題看到 ✅ 就整個畫面截一張，要看得到 ✅ 那幾行（組別、執行編號、驗證碼）和時間。缺截圖的題目不計分。
2. **總分卡截圖 × 1**：按最後一格「收工：總分卡」，整個畫面截圖。10 張截圖的執行編號要一樣、驗證碼要和總分卡對得上。

不用交筆記本連結，也不用下載筆記本。

| 狀況 | 原因 | 怎麼處理 |
|---|---|---|
| 按 ▶ 跳出「這個筆記本不是 Google 建立的」 | 從 GitHub 打開的筆記本都會問 | 按「仍要執行」 |
| 印出「先回到最上面的第 1 題」 | 還沒按第 1 題，或執行階段重新開始過 | 第 1 題的「接續碼」填執行編號，再按 ▶；往下每題再按一次 ▶ |
| 印出「要用到第幾題的結果」 | 後面的題目會用到前面的答案 | 先把那一題做到打勾 |
| 不小心重新整理或關掉網頁 | 沒有存副本，填過的答案不會留下 | 第 1 題接續碼填執行編號，資料一樣；答案要重填 |
| 明明填了數字，卻說「要填數字」 | 用了全形符號或多打了字 | 只填半形數字和小數點，例如 0.25 |
